Concept 2.5 — Iterables, Iterators & Generators



Iterable: An iterable is any Python object, such as a list, tuple, string, or range, that can be traversed one item at a time. An iterable can provide an iterator when passed to iter().
Iterator: An iterator is a stateful object that remembers its current position during iteration. It provides the next item whenever next(iterator) is called. When no elements remain, next() raises a StopIteration exception.
Generator: A generator is a special type of iterator produced by a generator function containing yield. It generates values lazily, one at a time, and pauses and resumes its execution state around each yield. Generators are particularly useful for memory-efficient processing and streaming patterns.

| Concept       | Memory Behavior                                         | State Tracking                                         | How It Is Created                                           |
| ------------- | ------------------------------------------------------- | ------------------------------------------------------ | ----------------------------------------------------------- |
| **Iterable**  | May contain/store its elements, depending on the object | Does not itself represent a current iteration position | Lists, tuples, strings, sets, ranges, etc.                  |
| **Iterator**  | Produces items one at a time                            | Remembers current position; advances via `next()`      | Created with `iter(iterable)` or implemented as an iterator |
| **Generator** | Computes values lazily on demand                        | Pauses and resumes function state at `yield`           | Function containing `yield`                                 |

Concept 2 — Basic structure of a decorator

Decorator: A decorator is a function that takes another function (or class) and returns a modified version, wrapping it with extra behaviour without changing its original code. The @decorator syntax applies it to the function defined beneath it.

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        result = func()
        print("Something is happening after the function is called.")
        return result
    return wrapper


def greet():
    return "Hello, World!"


decorated = my_decorator(greet)

print(decorated())
```

Concept 3 — @ syntax (clean shortcut)

```python
@my_decorator
def greet():
    return "Hello, World!"
```

```python
greet()
```

```python
def line_decorator(func):
    """This is a decorator that adds a line of equal signs before and the function call. """
    def wrapper():
        print("===============")
        result = func()
        print(result)
        print("===============")

    return wrapper    
```

```python
@line_decorator
def greet():
    return "How are you?"
```

```python
greet()
```

```python
@line_decorator
def rani():
    return "I love my husband."
```

```python
rani()
```

Concept 4 — Wrapping functions with arguments (*args, **kwargs)

```python
import time
```

```python
time.time()
```

```python
import functools
```

```python
def time(func):
    """This is a decorator that calculates the time taken to execute a function. """
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        import time
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"Time taken to exicute function {func.__name__}: {end_time - start_time} second")
        print(f"The function {func.__name__} use for {func.__doc__}")
        return result
    return wrapper
```

```python
@time
def addition(a, b):
    """This is a function that adds two numbers."""
    return a + b
```

```python
addition(1,3)
```

```python
addition.__name__
```

```python
addition.__doc__
```

```python
@time
def multiplication(a, b, c):
    return a * b * c
```

```python
multiplication(1, 2, 3)
```

Concept 1 — What does a context manager do?

Context manager: A context manager is an object used with the with statement that runs setup code on entry and guaranteed cleanup code on exit, even if an error occurs. It is commonly created with the @contextmanager decorator, where code before yield is setup and code after it is cleanup.

Concept 2 — Create your own using contextmanager

```python
from contextlib import contextmanager
import time
```

```python
@contextmanager
def timer(label):
    start_time = time.time()
    print(f"Starting {label}...")
    print("Starting time is", start_time)
    yield
    end_time = time.time()
    print(f"{label} took {end_time - start_time} second")
```

```python
with timer("my task"):
    total = sum(range(1000))
    print("Total sum is", total)
```

Dataclass

A dataclass is a class decorated with @dataclass that automatically generates boilerplate methods such as __init__, __repr__, and __eq__ from its type-annotated attributes.

Concept 3 — functools.lru_cache (Remember Results)

lru_cache

lru_cache is a decorator that caches, or memoizes, a function's return values based on its arguments.

When the same arguments are provided again, the stored result is returned instead of recomputing the result.
This avoids repeated expensive work.

```python
from functools import lru_cache
```

```python
@lru_cache(maxsize=None)
def slow_square(n):
    print(f"Calculating {n} square.....")
    return n **2
```

```python
slow_square(5)
```

```python
slow_square(5)
```

Concept 6 — Protocol (First Pass — Duck Typing)

Protocol

A Protocol defines an interface through the methods and attributes that an object must have, rather than requiring the object to inherit from a particular class.

Any object that provides the required members satisfies the protocol.

This is called structural typing, or explicit duck typing.

💡 Simple Explanation

A Protocol basically says:

“I don't care which class this object belongs to. It only needs to have this required method.”

```python
from typing import Protocol
```

```python
class Runnable(Protocol):
    def run(self) -> str:...

def execute(tool: Runnable) -> str:
    return tool.run()  
```

```python
class SearchTool:
    def run(self) -> str:
        return "Searching....."
```

```python
tool = SearchTool()

execute(tool)
```

🎤 Industry Interview Questions — Week 11
Real interview-style questions covering this week's topics, with model answers (in English). Try to answer them yourself first, then read the solution.

Q1. What is a generator, how does it differ from a list, and why does it matter for LLMs?

A generator (a function using yield, or a generator expression) produces values lazily, one at a time, pausing and resuming its state between values, instead of building the whole sequence in memory like a list. This means constant memory usage even over huge or infinite streams. It maps directly onto LLM streaming: tokens arrive and are yielded one by one so the UI can display them in real time, rather than waiting for the whole response.

Q2. What is a decorator and how does it work?

A decorator is a function that takes another function and returns a new function that wraps it, adding behavior before/after the original without changing its code. @my_decorator above a def is just syntactic sugar for func = my_decorator(func). Decorators use closures and typically *args, **kwargs to forward arguments, plus functools.wraps to preserve the original's name/docstring. Common examples: @retry, @timer, @lru_cache, and the @tool decorators used in agent frameworks.

Q3. What does @dataclass give you?

@dataclass auto-generates boilerplate for classes that mainly hold data: __init__, __repr__, and __eq__ based on the fields you declare with type hints. You get clean, readable, comparable objects with far less code, plus features like default values, frozen=True for immutability, and field(default_factory=...) for safe mutable defaults. It's ideal for modeling messages, tool definitions, and config objects.

Q4. What is functools.lru_cache and what are its constraints?

@lru_cache memoizes a function: it stores results keyed by the arguments so repeated calls with the same inputs return instantly instead of recomputing. It's great for expensive pure functions (and to save money by not re-calling an LLM/embedding for identical inputs). Constraints: arguments must be hashable (no lists/dicts), and you must not cache functions whose result should change over time or per call (current time, random values, live data) — you'd serve stale results.

Q5. In typing, what's the difference between Optional, Union, and a Protocol?

Optional[X] means "an X or None" (shorthand for Union[X, None]). Union[A, B] (or A | B) means the value can be one of several types. A Protocol defines structural typing ("duck typing" made explicit): any object that has the required methods/attributes satisfies the protocol, without needing to inherit from it — unlike an ABC, which requires explicit subclassing. Protocols are great for typing "anything with a .run() method."
