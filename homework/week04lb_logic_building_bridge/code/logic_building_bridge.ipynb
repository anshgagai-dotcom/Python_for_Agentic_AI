# Logic Building


## 5 - Step Method

1. Restate - Restating is expressing the problem in your own words to ensure you understand what needs to be solved.

2. Example First - Example First is the process of solving a small, concrete example manually to understand the pattern and expected behavior before writing any logic.

3. Pseudocode - Pseudocode is a step-by-step plan for solving a problem written in plain language instead of programming code.

4. Translate - Translation is converting pseudocode into actual Python code.

5. Trace (Dry Run) - Tracing is manually executing code line by line to verify its correctness and understand its behavior.

# LOGIC CLASS 2 — Pseudocode(Think in English before writing Python).

Pseudocode is a language-independent, step-by-step description of an algorithm written in plain English (or any natural language) before writing actual programming code.

Concept 1 — Find the Biggest Among 3 Numbers

Read the Question
        ↓
Restate the Problem
        ↓
Solve a Small Example
        ↓
Write Pseudocode
        ↓
Translate to Python
        ↓
Trace the Code
        ↓
Run and Check

Think in English first. Code in Python second.

Step 1 — Restate: You are given 3 numbers. Print the largest (biggest) number.
Mujhe 3 numbers milenge, mujhe sabse bada print karna hai.

Step 2 — Example: 
 a=4, b=9, c=2 → answer 9. (Kaagaz par: 4 aur 9 mein 9 bada, 9 aur 2 mein 9 bada → 9.)

Step 3 — Pseudocode (English, no Python):

1. maano sabse bada = a
2. agar b sabse bade se bada hai, toh sabse bada = b
3. agar c sabse bade se bada hai, toh sabse bada = c
4. sabse bada print karo

Step 4 — Translate (har line → Python):

```python
a, b, c = 4, 9, 3
biggest = a
if b > biggest:
    biggest = b
if c > biggest:
    biggest = c
print(biggest)
```

Step 5 — Trace: biggest: 4 → (b=9>4) 9 → (c=2>9? nahi) 9 → print 9. 

Concept 2 — Even or Odd from 1 to 10

Pseudocode:
Start from 1.

Go up to 10.

For each number

    If the number is divisible by 2

        Print the number and "Even"

    Otherwise

        Print the number and "Odd"

Translate: Python Code

```python
for n in range(1, 11):
    if n % 2 == 0:
        print(f"{n} Even")
    else:
        print(f"{n} Odd")
```
