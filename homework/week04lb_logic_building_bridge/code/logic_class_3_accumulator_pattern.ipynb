Q1 — Find the Product of All Numbers/ Homework 1 — Find the Product of All Numbers

Step 1: Restate

Find the multiplication of all numbers in the list.

Step 2: Example
        [2, 3, 4]

Step 3: Pseudocode
            1. Create a variable product and set it to 1.
            2. Loop through each number in the list.
            3. Multiply product by the current number.
            4. Store the result back in product.
            5. Print product.

```python
#Step 4: Code  Translate

nums = [2, 3, 4]
product = 1

for n in nums:
    product = product * n

print(product)
```

        | Loop  | n | product before | product after |
        | ----- | - | -------------- | ------------- |
        | Start | - | 1              | 1             |
        | 1     | 2 | 1              | 2             |
        | 2     | 3 | 2              | 6             |
        | 3     | 4 | 6              | 24            |


Homework Q2 — Count Words in a Sentence/ Ek sentence mein kitne words hain ginno (hint: .split()).

Step 1: Restate

Count how many words are present in the sentence.

Step 2: Example

Sentence:

I am learning to code

Words:

I
am
learning
to
code

Total:

5

Step 3: Pseudocode
            1. Split the sentence into words.
            2. Create count = 0.
            3. Loop through every word.
            4. Increase count by 1.
            5. Print count.

```python
# Step 4: Code Translate

sentence = "I am learning to code"

words = sentence.split()

count = 0

for w in words:
    count = count + 1

print(count)
```

Homework Q3 — Find the Smallest Number/ Ek list mein sabse chhota number dhoondho (bina min()).

Step 1: Restate

Find the smallest number without using min().

Step 2: Example

Input:

[8, 3, 9, 1, 5]

Smallest number:

1

Step 3: Pseudocode

1. Assume the first number is the smallest.
2. Loop through every number.
3. If the current number is smaller,update smallest.
4. Continue until the loop ends.
5. Print smallest.

```python
# Step 4: Code Translate

nums = [8, 3, 9, 1, 5]

smallest = nums[0]

for n in nums:
    if n < smallest:
        smallest = n

print(smallest)
```

| 3     | 9 | 3               | No            | 3              |
| 4     | 1 | 3               | Yes           | 1              |
| 5     | 5 | 1               | No            | 1              |


Homework Q4 — Reverse a String/ "hello" ko reverse karke naya string banao (accumulator = "").

Step 1: Restate

Reverse the word using an accumulator.

Step 2: Example

Input:

hello

Output:

olleh

Step 3: Pseudocode

1. Create an empty string.
2. Loop through every letter.
3. Add the current letter to the beginning of the new string.
4. Continue until all letters are processed.
5. Print the reversed string.

```python
# Step 4: Code Translate

word = "hello"

reversed_word = ""

for letter in word:
    reversed_word = letter + reversed_word

print(reversed_word)
```

| 5     | o      | `"lleh"`             | `"olleh"`           |
