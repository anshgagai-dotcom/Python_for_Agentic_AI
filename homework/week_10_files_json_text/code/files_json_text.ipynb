## CLASS 56 — Reading & Writing Files


Concept 1 — Writing to a File (with open)


 File handling with open():
open() returns a file object connected to a file on the disk. We tell Python which mode we want to use, such as reading or writing. When we use open() with a with statement, Python automatically closes the file when we are finished with it

```python
with open("my_first_file.txt", "w") as f:
    text = "Hello, You are my first file."
    f.write(text)
```

Concept 2 — File Modes (w vs a vs r)

📖 Technical definition — File mode:
A file mode is the string we give to open() that tells Python what we want to do with the file.

"r" means read
"w" means write and replace
"a" means append/add to the end

| Mode  | Meaning                    | What happens?                                           |
| ----- | -------------------------- | ------------------------------------------------------- |
| `"r"` | **Read** — read the file   | File must already exist, otherwise you get an error     |
| `"w"` | **Write** — write new data | ⚠️ **Old content is deleted/replaced**                  |
| `"a"` | **Append** — add data      | Old content stays safe and new data is added at the end |

Concept 4 — SECURITY: Avoid Path Traversal

Path traversal is a security vulnerability where untrusted input containing sequences like ../ is used to escape an intended directory and access files somewhere else on the system. It is prevented by resolving the final absolute path and checking that it stays inside an allowed base folder.

```python
from pathlib import Path
base_folder = Path("utils").resolve()
file_name = input("Enter the file you want to use from utils folder:")
file_path = (base_folder/file_name).resolve()

if file_path.is_relative_to(base_folder):
    data = file_path.read_text()
    print(data)
else:
    print("Access denied")
```

## CLASS 58 — JSON

Concept 1 — What Does JSON Look Like?

JSON:
JSON (JavaScript Object Notation) is a lightweight, text-based data format for storing and exchanging structured data as key-value objects and arrays. It is language-independent and is widely used by web and LLM APIs.

```python
## JSON Data

{
    "name": "Asha",
    "age": 17,
    "subject": ["Math", "Science"],
    "is_student": true
}
```

```python
## Python Dictionary

{
    "name": "Asha",
    "age": 17,
    "subject": ["Math", "Science"],
    "is_student": True
}
```

| Python                                                         | JSON                               |
| -------------------------------------------------------------- | ---------------------------------- |
| `True`                                                         | `true`                             |
| `False`                                                        | `false`                            |
| `None`                                                         | `null`                             |
| Usually single or double quotes can be used for Python strings | JSON strings use **double quotes** |


Concept 2 — Python → JSON String (dumps)

 — Serialization / Deserialization:
Serialization converts an in-memory Python object into JSON text. json.dumps() converts it to a string, while json.dump() writes it to a file.

Deserialization does the reverse. json.loads() converts JSON text into Python objects, while json.load() reads JSON from a file.

```python
import json
```

```python
data = {
    "name": "Asha",
    "age": 17,
    "subject": ["Math", "Science"],
    "is_student": True
}

json_string = json.dumps(data)
```

Concept 4 — JSON with a File (dump / load)

```python
data = {
    "name": "Asha",
    "age": 17,
    "subject": ["Math", "Science"],
    "is_student": True
}

with open("student_record.json", "w") as f:
    json.dump(data, f, indent=4)
```

```python
with open("student_record.json", "r") as f:
    data_from_file = json.load(f)
```

```python
data_from_file
```

```python
data_from_file["subject"]
```

| Function                | Meaning                  |
| ----------------------- | ------------------------ |
| `json.dumps(data)`      | Python → JSON **string** |
| `json.loads(text)`      | JSON **string** → Python |
| `json.dump(data, file)` | Python → JSON **file**   |
| `json.load(file)`       | JSON **file** → Python   |

Concept 1 — CSV (Comma-Separated Values)

 — CSV:
CSV (Comma-Separated Values) is a plain-text table format where each line represents a row and columns are separated by commas. Python's csv module can read and write CSV files, and DictReader can map each row to a dictionary using the header names.

```python
a, b, c,
1, 2, 3,
4, 5, 6,


[
    {"a": 1, "b": 2, "c": 3},
    {"a": 4, "b": 5, "c": 6}
]
```

```python
import csv
from typing_extensions import Writer

rows = [
    ["Name", "Marks"],
    ["Ansh", 85],
    ["Rahul", 90]
]

with open("marks.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(rows)
```

```python
with open("marks.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

Concept 3 — Regex (Finding Patterns)

— Regular Expression (Regex):
A regular expression is a compact pattern language used for matching, searching, and replacing text. Python's re module provides functions such as search() for the first match, findall() for all matches, and sub() for replacing matches

```python
import re
```

```python
text = """
 My contact : email >> anshgag.ai@gmail.com,
 and my mobile number is >> 7078743687
"""

numbers = re.findall(r"\d+", text)
print(numbers)
```

```python
numbers = re.search(r"\d+", text)
print(numbers.group())
```

```python
list_words = text.split(" ")
```

```python
def get_email(text):
    list_words = text.split(" ")
    for word in list_words:
     if "@" in word:
        print(word)

get_email(text)        
```

```python
email = re.search(r"\w+@\w+\.\w+", text)
print(email.group())
```

Concept 4 — Common Regex Patterns

```python
import re

text = "I have 3 cats and 12 dogs. #pets #animals"

print(re.findall(r"\d+", text))
print(re.findall(r"#\w+", text))
print(re.sub(r"\d+", "N", text))
```

| Pattern   | Meaning                             | Example                 |
| --------- | ----------------------------------- | ----------------------- |
| `\d`      | One digit (0–9)                     | `"5"`                   |
| `\D`      | Not a digit                         | `"R"`, `"@"`            |
| `\w`      | Word character: letter, digit, `_`  | `"user_1"`              |
| `\W`      | Non-word character                  | `"@"`, `"#"`, `"!"`     |
| `\s`      | Whitespace                          | space, tab, newline     |
| `\S`      | Non-whitespace                      | any non-space character |
| `.`       | Any single character except newline | `"a"`, `"9"`            |
| `[abc]`   | `a`, `b`, or `c`                    | `"a"`                   |
| `[^abc]`  | Anything except `a`, `b`, `c`       | `"p"`                   |
| `[a-z]`   | Lowercase letters                   | `a-z`                   |
| `[A-Z]`   | Uppercase letters                   | `A-Z`                   |
| `+`       | 1 or more times                     | `\d+`                   |
| `*`       | 0 or more times                     | `\d*`                   |
| `?`       | 0 or 1 time                         | `colou?r`               |
| `{n}`     | Exactly `n` times                   | `\d{4}`                 |
| `{n,m}`   | Between `n` and `m` times           | `\d{2,4}`               |
| `^`       | Start of string/line                | `^Hello`                |
| `$`       | End of string/line                  | `end$`                  |
| `\b`      | Word boundary                       | `\bcat\b`               |
| `(...)`   | Capturing group                     | `(\d{3})-(\d{4})`       |
| `(?:...)` | Non-capturing group                 | `(?:https\|http)`       |
| `a\|b`    | Either `a` OR `b`                   | `cat\|dog`              |

Concept 1 — subprocess (Running Commands)

 — subprocess:
The subprocess module runs external programs from Python as separate processes and can capture their output and exit code. Passing commands as a list of arguments instead of a shell string avoids shell interpretation and reduces command-injection risks.

```python
import subprocess
```

```python
result = subprocess.run(["python", "--version"], capture_output=True, text=True)
```

Concept 3 — Logging (Better Than print)

 — Logging:
Logging is the practice of recording structured, timestamped messages about a program's execution at severity levels such as INFO, WARNING, and ERROR. Unlike print, logging can be filtered by level and sent to files or other destinations. Sensitive information must never be logged.

```python
import logging
```

```python

logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")
```

```python
logging.info("Program started")
```

```python
logging.info("Program started")
logging.warning("Low memory")
logging.error("Something failed or no space to store message")
```
