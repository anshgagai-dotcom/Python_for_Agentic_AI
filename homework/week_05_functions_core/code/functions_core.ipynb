Concept 1 – Default Parameter Values (Optional Parameters)

A default parameter value is a value assigned to a parameter in the function definition. If the caller does not provide an argument for that parameter, Python automatically uses the default value, making the parameter optional.

Rule: Parameters with default values must come after parameters without default values.

# CLASS 33 – *args and **kwargs

Concept 1 – *args (Variable Number of Positional Arguments)

*args is a parameter prefixed with a single asterisk (*) that collects any number of positional arguments into a tuple. This allows a function to accept zero, one, or many positional inputs.

Note: args is just a common name. You can use any valid variable name (for example, *numbers, *values, *items). The * is what gives it this special behavior.

Why Do We Use *args?

Sometimes we don't know how many values the user will provide.

Examples:

Add many numbers
Calculate average
Print many names
Store multiple items

```python
def sum_all(*a):
    print("type of a: ", type(a))
    print("Values store in a: ", a)
    return sum(a)
```

```python
sum_all(1,2,3,47,18,19,20)
```

```python
total_sum = sum_all(1,2)
print("Sum of all arguments:", total_sum)
```

Concept 2 – **kwargs (Variable Number of Keyword Arguments)

**kwargs is a parameter prefixed with two asterisks (**) that collects any number of keyword arguments into a dictionary, where each keyword becomes a key and its value becomes the dictionary value.

Like args, kwargs is only a naming convention. You may use any valid name, such as **details or **settings. The ** gives it its special meaning.

Why Do We Use **kwargs?

When we don't know

how many keyword values will come
or what their names will be

```python
def my_bio(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
my_bio(name = "John", age = 30, city = "New York")
```

```python
my_bio(name = "John", age = 30, city = "New York", education = "MBA")

```

```python


def students_info(paper_marks, *marks, **info):
    total_marks = sum(marks)
    percentage = total_marks/(len(marks) *paper_marks) *100

    print("Student Information")

    for key, value in info.items():
        print(f"{key}: {value}")

    print("Total_marks: ", total_marks)
    print("Percentage : ", percentage)
```

# CLASS 34 – The Mutable Default Trap & Sentinels

Concept 1 – The Mutable Default Argument Trap

A mutable default argument is a mutable object (such as a list, dictionary, or set) used as a default parameter value in a function.

Python creates this default object only once, when the function is defined, and then reuses the same object every time the function is called without that argument.

As a result, any changes made to that object remain for future function calls, causing unexpected behavior.

```python
def add_item(item, cart = []):
    cart.append(item)
    return cart
```

```python
add_item(("mango"))
```

```python
add_item("apple")
```

```python
add_item("lemon")
```

Concept 2 – The Correct Solution (Using None)

Instead of using a mutable object as the default value, use None.

Inside the function, check whether the parameter is None.

If it is, create a brand-new object.

This ensures every function call gets its own fresh object.

```python
def add_item(item, cart = None):
    if cart is None:
        cart = []
    cart.append(item)
    return cart
```

```python
add_item("mango")
```

```python
add_item("apple")
```

Concept 3 – Sentinel Objects

A sentinel is a unique object used to represent "no value was supplied."

```python
object()
```

```python
type(object())
```

```python
__MISSING = object()

def update(value = __MISSING):
    if value is __MISSING:
        print("No value was given at all")
    elif value is None:
        print("User passed None value directly")
    else:
        print(f"User gave: {value}")        
```

```python
update()
```

```python
update(None)
```

```python
update(9)
```

🎤 Industry Interview Questions — Week 5
Real interview-style questions covering this week's topics, with model answers (in English). Try to answer them yourself first, then read the solution.

Q1. What is the difference between return and print in a function?

print sends text to the console for a human to read; it does not give the value back to the calling code. return hands a value back to the caller so it can be stored, reused, or passed on. A function that only prints returns None, so total = add(2, 3) would set total to None if add used print instead of return. Real, reusable functions (and every agent "tool") must return their result.

Q2. Why must parameters with default values come last, and what is the mutable default argument trap?

Defaults must follow non-default parameters because otherwise Python can't tell which positional argument fills which parameter. The famous trap is using a mutable default like def add_item(x, items=[]): that list is created once when the function is defined and shared across all calls, so it accumulates values between calls. The fix is the sentinel pattern: def add_item(x, items=None): items = [] if items is None else items.

Q3. What do *args and **kwargs do, and what order must parameters be in?

*args collects any extra positional arguments into a tuple; **kwargs collects any extra keyword arguments into a dict. This lets a function accept a variable number of inputs. The required order is: normal/positional parameters, then *args, then keyword-only parameters, then **kwargs. They're heavily used for wrappers and decorators that must forward arbitrary arguments.

Q4. What is the difference between positional and keyword arguments, and why prefer keyword arguments?

Positional arguments are matched by their position (greet("Asha", 17)); keyword arguments are matched by name (greet(name="Asha", age=17)). Keyword arguments are more readable, order-independent, and self-documenting — especially with boolean flags where send(retry=True) is far clearer than send(True). In production APIs, keyword arguments also make it safe to add or reorder parameters later.

Q5. How are arguments passed in Python — by value or by reference?

Python uses "pass by object reference" (sometimes called pass-by-assignment): the function receives a reference to the same object. If the object is immutable (int, str, tuple), it appears to behave like pass-by-value — reassigning inside the function doesn't affect the caller. If the object is mutable (list, dict), mutating it in place (e.g. lst.append(x)) is visible to the caller, while rebinding the name (lst = [...]) is not. Knowing this prevents accidental shared-state bugs.
