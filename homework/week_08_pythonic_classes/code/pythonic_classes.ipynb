Concept 2 — __str__


__str__ is a special method that defines the human-readable string representation of an object.

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Dog: {self.name}, {self.age} year old"

d = Dog("Buddy",  3)
print(d)  
```

Concept 3 — __repr__

__repr__ is a special method that defines a developer-oriented representation of an object, mainly useful for debugging and inspection.

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Dog: {self.name}, {self.age} year old"
    
    def __repr__(self):
        return f"Dog(name = {self.name}, age = {self.age})"


d = Dog("Buddy",  3)
print(d)  
```

Concept 2 — __eq__

__eq__ is a special method that defines how the == operator determines whether two objects are equal.

Simple meaning
__eq__ means:

"Python, this is how I want you to compare my objects."

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

p1 = Point(1, 2)
p2 = Point(1, 2)
p1 == p2            
```

```python

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, data):
        return self.x == data.x and self.y == data.y

p1 = Point(1, 2)
p2 = Point(1, 2)
p1 == p2    
```

```python
p3 = Point(3, 4)
p1 == p3
```

## CLASS 48 — classmethod & staticmethod

Concept 1 - @classmethod Alternative Constructor

An alternative constructor is a class method that provides another way of creating an object from different input data.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_string(cls, data):
        name, age = data.split(",")
        return cls(name, int(age))
```

```python
p1 = Person("Raju", 55)
```

```python
p1.name, p1.age
```

```python
p2 = Person.from_string("Raju, 55")
```

```python
p2.name, p2.age
```

Concept 2 — from_dict() Class Method

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_dict(cls, data):
        name, age = data["name"], data["age"]
        return cls(name, age)

    @classmethod
    def from_string(cls, data):
        name, age = data.split(",")
        return cls(name, int(age))    

    
```

```python
p1 = Person.from_dict({"name": "Raju", "age": 55})
p1.name, p1.age
```

## Combination

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_dict(cls, data):
        name, age = data["name"], data["age"]
        return cls(name, age)

    @classmethod
    def from_string(cls, data):
        name, age = data.split(",")
        return cls(name, int(age))    

    def __repr__(self):
        return f"Person(name={self.name}, age={self.age})"

    def __str__(self):
        return f"Person: {self.name}, {self.age} year old"
            
```

Concept 3 — What is @staticmethod?



A static method is a method that does not receive the instance (self) or the class (cls) automatically. It behaves like a normal function but is placed inside a class because it is logically related to that class.

```python
class MathHelper:
    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod
    def multiply(a, b):
        return a * b    
```

```python
MathHelper.add(1, 3)
```

```python
MathHelper.multiply(1, 3)
```

| Method Type     | First Parameter | Needs Object Data? | Needs Class Data? | Common Use               |
| --------------- | --------------- | ------------------ | ----------------- | ------------------------ |
| Normal method   | `self`          | Yes                | Can access class  | Work with an object      |
| `@classmethod`  | `cls`           | No                 | Yes               | Alternative constructors |
| `@staticmethod` | None            | No                 | No                | Utility/helper functions |


## CLASS 49 — property

Concept 1 — Computed Property

@property is a Python decorator that allows us to use a method like an attribute, without using parentheses ().

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius
    @property
    def area(self):
        return 3.14 * self.radius ** 2    
```

```python
circle = Circle(4)

circle.area
```

```python
circle.radius = 10
circle.area
```

```python
class Temperature:
    def __init__(self, celsius=0):
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Below absolute zero!")

        self._celsius = value
        
                    
```

```python
temp = Temperature()
```

```python
temp.celsius
```

```python
temp.celsius = 6
```

```python
temp.celsius
```

```python
print(temp.celsius)
```

```python
temp.celsius = -275
```

## CLASS 50 — Composition

Concept 1 — Composition = Object Inside Another Object


Composition is a design approach where one object is built using other objects as its attributes.

```python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        return f"{self.horsepower} HP Engine Started"


class Car:
    def __init__(self, brand, horsepower):
        self.brand = brand
        self.horsepower = horsepower
        self.engine = Engine(self.horsepower).start()

    def start(self):
        return self.engine    
```

```python
Engine(4).start()
```

```python
Car("Tata", 1500).start()
```

## Class inside class

```python

class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        return f"{self.horsepower} HP Engine Started"


class Car:
    class Features:
        def __init__(self, color):
            self.color = color

    def __init__(self, brand, horsepower):
        self.brand = brand
        self.horsepower = horsepower
        self.engine = Engine(self.horsepower).start()
        self.color = self.color()

    def start(self):
        return self.engine

    def color(self, color="red"):
        return self.Features(color).color
```

```python
Car("Tata", 1300).color
```

Concept 2 — IS-A vs HAS-A

| **Inheritance (IS-A)**                              | **Composition (HAS-A)**                              |
| --------------------------------------------------- | ---------------------------------------------------- |
| **Dog IS-A Animal**                                 | **Car HAS-A Engine**                                 |
| Describes a **type relationship**                   | Describes a **parts relationship**                   |
| `class Dog(Animal)`                                 | `self.engine = Engine()`                             |
| **Rule:** If X is a type of Y → use **Inheritance** | **Rule:** If X has Y as a part → use **Composition** |
| Example: A Dog **is an** Animal                     | Example: A Car **has an** Engine                     |


🎤 Industry Interview Questions — Week 8
Real interview-style questions covering this week's topics, with model answers (in English). Try to answer them yourself first, then read the solution.

Q1. What is the difference between __str__ and __repr__?

__str__ returns a human-friendly, readable string (used by print() and str()). __repr__ returns an unambiguous, developer-facing string meant for debugging — ideally one that looks like valid code to recreate the object, e.g. Message(role='user', content='hi'). repr() and the interactive prompt use __repr__, and containers (like lists) show the __repr__ of their elements. If you only implement one, implement __repr__, because Python falls back to it when __str__ is missing.

Q2. What is the difference between a class attribute and an instance attribute, and what's the common pitfall?

A class attribute is defined on the class and shared by all instances; an instance attribute is set per object (usually in __init__) and belongs only to that object. The pitfall is a mutable class attribute, like tools = [] at class level — every instance shares the same list, so appending via one instance affects all of them. Mutable per-object state must be created inside __init__.

Q3. When would you use a @classmethod, a @staticmethod, and a regular instance method?

An instance method (takes self) operates on a specific object's data. A @classmethod (takes cls) operates on the class itself — the classic use is an alternative constructor like Message.from_dict(data) that builds and returns an instance. A @staticmethod takes neither self nor cls; it's just a utility function grouped inside the class because it's logically related, but it doesn't use object or class state.

Q4. What is @property and why use it instead of a public attribute or explicit getter/setter methods?

@property lets a method be accessed like an attribute (obj.area instead of obj.area()), so you can start with a plain attribute and later add computed values or validation without changing the calling code. Compared to Java-style get_x()/set_x() methods, properties keep the clean attribute syntax while still allowing validation in the setter (e.g. rejecting a negative temperature). This is the Pythonic way to control attribute access.

Q5. Why implement __eq__, and what must you remember about __hash__?

By default two objects are "equal" only if they are the same object in memory. Implementing __eq__ lets you define value-based equality (two Message objects with the same fields are equal). Important: if you override __eq__, Python sets __hash__ to None, making instances unhashable (unusable in sets/dict keys). If you need them hashable, also implement __hash__ consistently with __eq__ — equal objects must have equal hashes.
