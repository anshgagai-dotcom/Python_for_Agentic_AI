## Class 52 - try / except / else / finally 

Concept 1 - basic try / except

try block: A block containing code that may raise an exception and therefore needs exception handling.

try = "Pehle yeh risky kaam karke dekho."



except block: A block that handles an exception raised in the associated try block.

except = "Agar error aaye, toh yeh kaam karo."

```python
try:
    pass
except:
    pass
else:
    pass
finally:
    pass
```

```python
try:
    print("Start of program")
    result = 10/0
    print("end of program")
except ZeroDivisionError:
    print("Error: Division by zero")    
```

Concept 2 - as e

as e binds the caught exception object to the variable e.


Error ko ek variable mein pakad lena.

```python
try:
    print("Start of program")
    result = 10/0
    print("end of program")
except ZeroDivisionError as e:
    print(f"Error: {e}")    
```

```python
try:
    age = int(input("Enter your age: "))
    print(f"Your {age} years old")
except ValueError as e:
    print(f"Error: {e}")    
```

Concept 3 - else & finally

else block: An optional block that executes only when the try block completes without raising an exception.

else = Error nahi aaya → Success wala kaam karo.



finally block: An optional block that normally executes regardless of whether an exception occurred, commonly used for cleanup.

finally = Kuch bhi ho, main chalunga.

```python
try:
    number = int(input("Enter a number: "))
except ValueError as e:
    print(f"Error: {e}")
else:
    print(f"Well done, you have entered a valid number: {number}")
finally:
    print("See you next time")

```

## Class 53 - Specific Exceptions

Concept 1 - Multiple Exception Handlers

Multiple exception handlers are multiple except blocks attached to one try block, with each handler targeting a particular exception type.

Different errors ke liye different solutions.

```python
try:
    numbers = [1,2,3,4,5]
    index = int(input("enter an index: "))
    print(numbers[index])
except ValueError as e:
    print(f"Error: {e}")
except IndexError as e:
    print(f"Error: {e}")
```

Concept 2 - Python executes the first except handler whose exception type matches the raised exception.

Python line-by-line except check karta hai:

```python
try:
    number = [1,2,3,4,5]
    index = int(input("Enter an index: "))
    print(numbers[index])
except (ValueError,IndexError) as e:
    print(f"Error: {e}")    
```

Concept 3 - A tuple of exception types allows one except block to handle multiple specified exception types.


Do ya zyada errors ko ek hi handler se handle karna.

```python
try:
    number = [1,2,3,4,5]
    index = int(input("Enter an index: "))
    print(numbers[index])
except Exception as e:
    print(f"Error: {e}") 
```

# CLASS 54 — raise & CUSTOM EXCEPTIONS

## Concept 16 — raise & Custom Exception

raise is a Python statement used to deliberately trigger an exception when the programmer detects an invalid or unacceptable condition.



Custom Exception: A user-defined exception type created by subclassing Exception or one of its subclasses.


Custom Exception = Apne project ke liye apna special error banana.

```python
def set_age(age: int)->int:
    if (age < 0) or (age > 150):
        raise ValueError("Age can not be negative or more than 150")

    return age
```

```python
set_age(0)
```

```python
set_age(-9)
```

 Concept 3 — custom exception class

 
 Custom exception: A custom exception is a user-defined error type created by subclassing Exception (or one of its subclasses). It gives a project its own named, meaningful error types that can be raised and caught specifically.

```python
class InsufficientFundError(Exception):
    "Raised when withdrawl exceeds the balance"
    pass
```

```python
def withdraw_good(balance, amount):
    if amount > balance:
        raise InsufficientFundError(f"Can not withdraw {amount}, current balance is {balance}.")
    return balance - amount    
```

```python
withdraw_good(500, 1000)
```

```python
try:
    withdraw_good(500, 1000)
except InsufficientFundError as e:
    print(f"Error: {e}")    
```

Concept 4 — exception hierarchy 

```python
class ToolError(Exception):
    """Base error for all tool problems."""
    pass
```

```python
class ToolNotFoundError(ToolError):
    "XYZ tool not found"
    pass
```

```python
class ToolTimeoutError(ToolError):
    "Taking long time to get tool"
    pass
```

```python
try:
    tool_name = "Get current time tool"
    time_taken = "1000 seconds"
    raise ToolTimeoutError(f"Taking long time [{time_taken}] to  call {tool_name}")
except ToolTimeoutError as e:
    print(f"Error: {e}")    
```

# CLASS 55 — Retry Pattern (Project Class)

Concept 1 - Retry

Retry Pattern: The retry pattern re-attempts an operation that failed with a transient error, up to a maximum number of attempts.


Retry = Fail hua? Koi baat nahi. Dobara try karo.

```python
import time
```

```python
def flaky_task(attempt):
    if attempt < 3:
        raise ConnectionError("Network error")
    return "Success!"


max_attempt = 3 

for attempt in range(1, max_attempt+1):
    try:
        result = flaky_task(attempt)
        print(f"Attempt {attempt}: {result}")
        break
    except ConnectionError as e:
        print(f"Attempt {attempt} failed: {e}")

        if attempt < max_attempt:
            time.sleep(1)
        else:
            print("All attempts failed, Giving up.")
                
```

```python
flaky_task(4)
```

Concept 2 — Backoff

```python
for attempt in range(1, 6):
    wait = 2 ** (attempt - 1)
    time.sleep(wait)

    print(f"Attempt {attempt}, would wait {wait}s on failure")
```
