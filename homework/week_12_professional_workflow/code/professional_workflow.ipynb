## Class 66 - Virtual Environments & pip

Concept 1 — What are a Virtual Environment and pip, and Why Do We Need Them?

📖 Technical definition — Virtual environment and requirements.txt: A virtual environment is an isolated directory containing its own Python interpreter and installed packages for a single project. A requirements.txt file lists the project's package dependencies (and versions), so an identical environment can be recreated elsewhere with pip install -r requirements.txt.

📖 Technical definition — pip (Package Installer for Python): pip is the standard package manager for Python, used to install, update, and manage third-party packages from PyPI (Python Package Index). It comes pre-installed with standard Python distributions.

"A virtual environment (venv) is a separate, isolated Python setup for one project. Its packages stay within that project and remain separate from the rest of the system. This prevents projects from interfering with each other."

🛠️ pip Check & Upgrade (Before Environment Setup)

"When you install Python, pip usually comes with it. However, it is a good practice to check and upgrade it before creating a virtual environment:"

Check whether pip is working and see its version:
pip --version

Or:
python -m pip --version

Upgrade pip to the latest version (recommended):
python -m pip install --upgrade pip

If pip is missing (in system Python):
python -m ensurepip --default-pip

Concept 2 — Create and Activate a venv

Create a venv (one time)
python -m venv .venv

Activate it (every time you work on the project)
Windows (PowerShell):
.venv\Scripts\Activate.ps1

Mac/Linux:
source .venv/bin/activate

"python -m venv .venv creates a .venv folder (the project's isolated environment). activate turns it 'on' — now you will see (.venv) in the terminal, which means you are working inside the isolated environment. To deactivate it, simply type deactivate."

Note (Windows): If Activate.ps1 gives a permission error, run this once:

Set-ExecutionPolicy -Scope CurrentUser RemoteSigned

This gives PowerShell permission to run scripts. It is a common Windows issue.

Concept 3 — Install Packages Using pip

After activating the venv:
pip install requests
pip install pydantic
pip list                    # See what is installed

Concept 4 — requirements.txt (The Project's Shopping List)

Save all installed packages + versions into a file

pip freeze > requirements.txt


The file may look like:

pydantic==2.9.0
requests==2.32.0


"requirements.txt is your project's dependency list. Someone else (or you, on a new computer) can recreate the same setup simply by running:"


pip install -r requirements.txt


"The -r means 'install everything listed in this file'. This is the real solution to the 'works on my machine' problem — everyone can use the same package versions."


💻 Demo — New Project Setup
mkdir my_agent_project
cd my_agent_project
python -m venv .venv
.venv\Scripts\Activate.ps1        # Windows
pip install requests
pip freeze > requirements.txt

"These are the basic steps for starting every new Python project. This should become muscle memory."

## CLASS 67 — Modern Packaging & Lazy Imports

"requirements.txt is the traditional way of managing dependencies. Modern Python projects commonly use pyproject.toml — a file that can contain the project's metadata, dependencies, and tool configuration. Modern tools such as uv make dependency management much faster and simpler. Today we will get introduced to these concepts."

Concept 1 — pyproject.toml (The Project's ID Card)


— pyproject.toml: pyproject.toml is the standard configuration file for modern Python projects. It can declare project metadata, supported Python versions, dependencies, and tool configuration.


"pyproject.toml is a file that contains important information about your project: its name, version, description, Python requirement, dependencies, and other settings. It is the modern standard for Python project configuration."


[project]
name = "my-agent"
version = "0.1.0"
description = "My first AI agent project"
requires-python = ">=3.15"
dependencies = [
    "requests>=2.32",
    "pydantic>=2.9",
]


"Notice that the project name, version, Python requirement, and dependencies can all be defined in one place. This makes project configuration cleaner than maintaining separate legacy configuration files."

Concept 2 — uv (A Fast Modern Python Tool)

"uv is a modern, extremely fast Python package and project management tool written in Rust. It can create virtual environments, install packages, manage dependencies, and run Python projects."


📥 Installing uv

"You normally install uv once on your system:"

Method 1 — Install using pip:
pip install uv

Method 2 — Official standalone installer:

Windows (PowerShell):
powershell -c "irm https://astral.sh/uv/install.ps1 | iex"

Mac/Linux:
curl -LsSf https://astral.sh/uv/install.sh | sh

Check whether uv is installed:
uv --version

uv Workflows — Two Approaches

"uv supports two main styles of working. Both are valid, but for modern projects, the project-based workflow is generally preferred."


— uv workflows:

Project Workflow (uv init, uv add, uv sync) — Uses pyproject.toml and uv.lock to manage project dependencies and create reproducible environments.

Pip-Compatible Workflow (uv venv, uv pip install) — Provides fast alternatives to traditional venv and pip commands. It is useful for quick experiments and existing projects that use requirements.txt.

| Feature           | Official Project Workflow                      | Pip-Compatible Workflow                 |
| ----------------- | ---------------------------------------------- | --------------------------------------- |
| **Best For**      | New projects & production/company repositories | Legacy projects & quick experiments     |
| **Configuration** | `pyproject.toml` + `uv.lock`                   | `requirements.txt` or none              |
| **Venv Creation** | Automatically managed by `uv`                  | Manual using `uv venv`                  |
| **Add Packages**  | `uv add requests`                              | `uv pip install requests`               |
| **Run Code**      | `uv run python main.py`                        | Activate `.venv`, then `python main.py` |


Scenario B — Existing Company Project

"Suppose you join a company and clone an existing repository from GitHub."

Case 1 — Existing project uses requirements.txt
cd existing_company_project

Create a virtual environment:
uv venv

Windows:
.venv\Scripts\Activate.ps1

mac/Linux:
source .venv/bin/activate

Install dependencies:
uv pip install -r requirements.txt
Case 2 — Migrating to a Modern uv Project
cd existing_company_project

1. Create a basic pyproject.toml:
uv init --bare

2. Add dependencies from requirements.txt:
uv add -r requirements.txt

3. Create/update the environment and lock file:
uv sync

"The main idea is simple: uv can work like a faster alternative to venv + pip, while its project workflow integrates naturally with pyproject.toml and uv.lock."

Concept 3 — Semantic Versioning (>=, ==, ~=)

📖 Technical definition — Semantic versioning: Semantic versioning commonly represents releases as MAJOR.MINOR.PATCH. MAJOR versions may contain breaking changes, MINOR versions generally add backward-compatible functionality, and PATCH versions generally contain backward-compatible fixes.


🔢 Version Format

Example:

2.32.0

MAJOR (2) — potentially breaking changes

MINOR (32) — new backward-compatible features

PATCH (0) — bug fixes and small changes

Version Specifiers

"requests>=2.32" 
    
"requests==2.32.0"
   
"requests~=2.32"
     

"The important operators to remember are:"

>= → this version or newer

== → exactly this version

~= → compatible versions within the specified compatibility range

Concept 4 — Lazy Imports

📖 Technical definition — Lazy import: A lazy import delays importing a module until the point where it is actually needed, rather than importing it when the program initially starts.

"Normally, when Python encounters an import statement at the top of a file, it imports that module during program startup. If the module is large or expensive to initialize, this can increase startup time."

🔗 Agentic AI Connection

"Modern AI projects often use pyproject.toml together with tools such as uv or Poetry for clean dependency management.

Lazy imports are also useful in CLI-based AI agents because not every command needs every heavy AI library. Loading a dependency only when its functionality is required can improve startup performance.

These are important professional Python engineering skills because real-world AI projects are not only about writing models and prompts — they also require reliable project structure and dependency management."

## CLASS 68 — Project Structure

Concept 1 — What Does a Good Project Look Like?

A professional Python project can look like this:

my_agent_project/
├── .venv/                  # Virtual environment (NOT in Git)
├── src/                    # All source code
│   ├── __init__.py         # Package initialization
│   ├── tools.py            # Tool functions
│   ├── agent.py            # Agent class
│   └── main.py             # Entry point
├── tests/                  # Test files
│   └── test_tools.py
├── .env                    # Secrets (NOT in Git)
├── .gitignore              # Files Git should ignore
├── requirements.txt        # Or pyproject.toml
└── README.md               # Project documentation

"Notice that everything has its own place: code goes inside src/, tests inside tests/, secrets inside .env, and project information inside README.md. This separation makes professional projects easier to manage."

Concept 2 — __init__.py (Making a Directory a Package)

— Package and __init__.py: A package is a directory containing related Python modules. An __init__.py file can mark a directory as a Python package and can also contain package initialization code, exports, and public API definitions.

Concept 3 — README.md (The Face of Your Project)

"A README.md file is written in Markdown and provides an introduction and instructions for your project. When someone opens your project on GitHub, the README is one of the first things they see."

A simple REA

Concept 4 — .gitignore (What Git Should NOT Track)

— .gitignore: .gitignore is a text file containing patterns for files and directories that Git should not track. It is commonly used for virtual environments, generated files, caches, and secret files such as .env.

"Some files should not be committed to Git. For example, virtual environments are large and can be recreated, while .env files may contain API keys and other secrets."

## CLASS 69 — Git & GitHub

"Git is like a 'time machine' for your code — every commit is a save point that allows you to go back to an earlier version. GitHub is where you can store your code online for backup, collaboration, and portfolio purposes. Git is a daily tool for professional software engineers

Concept 1 — What Does Git Do?

"Git stores snapshots of your project called commits. If something goes wrong, you can inspect or return to earlier versions. If you want to experiment with a new feature, you can create a branch."

Git gives you the ability to:

Track changes

Create save points

Experiment safely

Work with other developers

Review project history


Concept 2 — Basic Git Workflow

— Commit and branch: A commit is a saved snapshot of tracked project files at a particular point in time. A branch is a separate line of development that allows changes to be developed independently before being merged into another branch.

The Basic Four Commands:

git init

git status

git add .

git commit -m "First commit"


What Does Each Command Do?

git init

Initializes the current folder as a Git repository.

git status

Shows which files are modified, staged, or untracked.

git add .

Stages your changes so they can be included in the next commit.

git commit -m "First commit"

Creates a snapshot of the staged change

Concept 3 — Write Good Commit Messages

✅ Good

git commit -m "Add calculator tool"


❌ Bad

git commit -m "stuff"

Good commit messages make your project history easier to understand months later.

Useful patterns include:

Add ...

Fix ...

Update ...

Remove ...

Refactor ...



Examples:

git commit -m "Add calculator tool"

git commit -m "Fix agent response handling"

git commit -m "Update README setup instructions"

Concept 4 — Git History and Branches

View Commit History

git log --oneline

This displays a compact list of previous commits.


Create a Branch

git branch feature-x

Switch to the Branch

git checkout feature-x



Or create and switch in one command:

git checkout -b feature-x

Think of it like this:

After creating an empty GitHub repository:

After the first push, you can usually use:

"Before pushing your project, ALWAYS check that sensitive files such as .env are ignored."

## CLASS 70 — Secrets & Profiling (Project Class)

Concept 1 — NEVER Put Secrets Directly in Code

❌ NEVER do this:

api_key = "sk-1234567890abcdef"

"Hard-coding an API key directly into your source code is a major security mistake. Your source code may eventually be committed to Git and pushed to GitHub. If the repository is public, the secret can be exposed."



Other examples of secrets include:

API keys

Database passwords

Access tokens

Private credentials

Cloud service credentials



❌ Bad Practice

OPENAI_API_KEY = "your-real-api-key"

DATABASE_PASSWORD = "my-secret-password"



✅ Better Practice

Keep secrets outside your source code and load them through environment variables.

Concept 2 — .env File

📖 Technical definition — Environment variables and .env: Environment variables are configuration values stored outside source code and made available to a running process. A .env file is a local file commonly used to store these values during development. A library such as python-dotenv can load them into the process environment.

"Keep your local development secrets in a separate .env file. Make sure .env is included in .gitignore so it is not committed to Git."

📄 .env
OPENAI_API_KEY=sk-your-real-key-here
DATABASE_URL=postgres://localhost/mydb

Important: The .env file is not automatically secure just because it has that name. Its security depends on how you protect it and whether you prevent it from being committed or exposed

Concept 3 — .env.example

"If .env contains real secrets, we cannot commit it. But another developer still needs to know which environment variables the project requires."

The solution is .env.example.

📄 .env.example
OPENAI_API_KEY=
DATABASE_URL=

This file contains:

Variable names
Empty placeholder values
No real secrets

It can be committed to Git.

Typical Workflow
.env.example
     ↓
Developer copies it
     ↓
.env
     ↓
Developer adds their own secrets

For example:

cp .env.example .env

Then the developer fills in their own values.

Concept 4 — Profiling: Find the Slow Parts

📖 Technical definition — Profiling: Profiling is the process of measuring where a program spends its execution time or memory so that optimization efforts can focus on actual bottlenecks.

"When your program is slow, don't guess which part is responsible. Measure it."

This gives us the important engineering rule:

Measure, Don't Guess.

🎤 Industry Interview Questions — Week 12
Q1. What is the difference between requirements.txt and pyproject.toml, and why pin versions?
Model Answer

requirements.txt is a simple dependency list commonly used with:

pip install -r requirements.txt

pyproject.toml is the modern Python project configuration standard. It can define project metadata, Python requirements, dependencies, and tool configuration.

We constrain or lock dependency versions to improve reproducibility. Without version constraints or a lock file, different environments may install different dependency versions, causing "works on my machine" problems.

Q2. Describe the basic Git workflow and what makes a good commit message.
Model Answer

The basic workflow is:

git add
   ↓
git commit
   ↓
git push

git add stages changes, git commit saves a snapshot to local Git history, and git push sends commits to a remote repository such as GitHub.

A good commit message is short, clear, and describes the change.

For example:

git commit -m "Add retry logic to LLM client"

Small, focused commits are generally easier to review, understand, and revert.

Q3. How do you keep secrets out of your code and Git?
Model Answer

I store secrets such as API keys and database passwords outside the source code, typically as environment variables. During local development, I can use a .env file with python-dotenv.

I add .env to .gitignore so it isn't committed, and I provide a .env.example containing the required variable names with empty values.

For example:

# .env.example
OPENAI_API_KEY=
DATABASE_URL=

Before committing, I check:

git status

If a real secret is accidentally committed, I immediately revoke or rotate it because removing the file later does not undo the exposure.

Q4. What is profiling, and when should you optimize?
Model Answer

Profiling measures where a program spends its execution time or resources. Python provides tools such as cProfile, while time.perf_counter() can be used for precise timing of specific operations.

The general approach is:

Make it correct
     ↓
Measure
     ↓
Find bottleneck
     ↓
Optimize
     ↓
Measure again

I avoid premature optimization because it can add unnecessary complexity without solving the actual performance problem.

Q5. What does a clean Python project structure look like, and why does __init__.py matter?
Model Answer

A clean project separates source code, tests, documentation, configuration, and dependencies.

For example:

my_project/
├── src/
│   ├── __init__.py
│   ├── tools.py
│   └── main.py
├── tests/
├── .gitignore
├── README.md
└── pyproject.toml

__init__.py can define a package's initialization and public interface and allows objects to be re-exported for cleaner imports.

For example:

from .tools import add
from .agent import Agent

__all__ = ["add", "Agent"]

Then:

from src import Agent, add

A clean structure makes a project easier to understand, test, maintain, package, and hand over to another developer.
