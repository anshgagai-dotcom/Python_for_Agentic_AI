Concept 3 – Return Statement

The return statement ends the execution of a function and sends a value back to the caller. The returned value can be stored, reused, or used in further calculations.

If no value is returned, Python automatically returns None.

```python
def sub(a, b):
    sub_result = a - b

    return sub_result
```

```python
sub(6,4)
```

```python
sub_final_result = sub(7,1)
```

```python
sub_final_result
```

```python
def sub(a, b):
    sub_result = a - b
    return print("Sub of a and b s ", sub_result)
```

```python
sub(8,9)
```

# CLASS 32 – Default Parameters & Keyword Arguments

Concept 1 – Default Parameter Values (Optional Parameters)

A default parameter value is a value assigned to a parameter in the function definition. If the caller does not provide an argument for that parameter, Python automatically uses the default value, making the parameter optional.

Rule: Parameters with default values must come after parameters without default values.

```python
def greet(user_name, greeting = "Hello"):
    print(f"{greeting}, {user_name}")
```

```python
greet("Laado")
```

```python
greet("Laado","Hi")
```

```python
greet("Laado", "How are you?")
```

Why Do We Use Default Parameters?

Default parameters make functions:

More flexible
Easier to use
Suitable for common cases without requiring every argument
Simpler to maintain

```python
greet("How are you?", "Ram")
```

Positional Arguments
Positional arguments are matched to parameters based on their position (order) in the function call.


Keyword Arguments
Keyword arguments are matched to parameters by explicitly specifying the parameter name using the syntax:

```python
def order(item, qty = 1, size = "medium"):
    print(f"Your order of {item} with quantity {qty} and size is {size}")
    
```

# CLASS 33 – *args and **kwargs

Concept 1 – *args (Variable Number of Positional Arguments)

*args is a parameter prefixed with a single asterisk (*) that collects any number of positional arguments into a tuple. This allows a function to accept zero, one, or many positional inputs.

Note: args is just a common name. You can use any valid variable name (for example, *numbers, *values, *items). The * is what gives it this special behavior.

Why Do We Use *args?

Sometimes we don't know how many values the user will provide.

Examples:

Add many numbers
Calculate average
Print many names
Store multiple items

```python
def sum_all(*a):
    print("type of a: ", type(a))
    print("Values store in a: ", a)
    return sum(a)
```

```python
sum_all(1,2,3,47,18,19,20)
```

```python
total_sum = sum_all(1,2)
print("Sum of all arguments:", total_sum)
```

Concept 2 – **kwargs (Variable Number of Keyword Arguments)

**kwargs is a parameter prefixed with two asterisks (**) that collects any number of keyword arguments into a dictionary, where each keyword becomes a key and its value becomes the dictionary value.

Like args, kwargs is only a naming convention. You may use any valid name, such as **details or **settings. The ** gives it its special meaning.

Why Do We Use **kwargs?

When we don't know

how many keyword values will come
or what their names will be

```python
def my_bio(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
my_bio(name = "John", age = 30, city = "New York")
```

```python
my_bio(name = "John", age = 30, city = "New York", education = "MBA")

```

```python
my_bio(name = "John", 
age = 30, 
city = "New York", 
education = "MBA", 
salary = 100000)

```

Concept 3 – Using Normal Parameters, *args, and **kwargs Together

```python


def students_info(paper_marks, *marks, **info):
    total_marks = sum(marks)
    percentage = total_marks/(len(marks) *paper_marks) *100

    print("Student Information")

    for key, value in info.items():
        print(f"{key}: {value}")

    print("Total_marks: ", total_marks)
    print("Percentage : ", percentage)
```

Class 33 Summary


Concept	Definition

*args	Collects any number of positional arguments into a tuple.

**kwargs	Collects any number of keyword arguments into a dictionary.

Tuple	An ordered, immutable collection created by *args.

Dictionary	A collection of key-value pairs created by **kwargs.

Order Rule	Normal parameters → *args → **kwargs.

Main Benefit	Makes functions flexible enough to accept a variable number of inputs.
