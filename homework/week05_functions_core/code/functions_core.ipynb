# WEEK 5 — Functions Core

Function: A function is a named, reusable block of code defined using the def keyword that performs a specific task. A function runs only when it is called. It can accept inputs (parameters), perform operations, and optionally return a result.

## Class 31: Defining & Calling Functions

Concept 1 – Defining and Calling Functions

A function is a named, reusable block of code defined using the def keyword that performs a specific task. A function executes only when it is called.

```python
def addition():
    a = 5
    b = 6
    print("addition of a and b is ", a + b)
    
```

```python
addition()
```

Concept 2 – Parameters and Arguments

Parameter
A parameter is a variable listed in the function definition that receives data when the function is called.

Argument
An argument is the actual value passed to the function during the function call.

```python
def addition(a, b, c):
    print("addition of a, b, and c is ", a + b + c)
    

```

```python
a = 5
b = 6
c = 8
addition(a, b, c)
```

```python
addition(6, 9, 1)
```

```python
addition(6, 8, c)
```

```python
def sub(a, b):
    print("Sub of a and b is ", a - b)
```

```python
sub(20, 15)
```

Concept 3 – Return Statement

The return statement ends the execution of a function and sends a value back to the caller. The returned value can be stored, reused, or used in further calculations.

If no value is returned, Python automatically returns None.

```python
def sub(a, b):
    sub_result = a - b

    return sub_result
```

```python
sub(6,4)
```

```python
sub_final_result = sub(7,1)
```

```python
sub_final_result
```

```python
def sub(a, b):
    sub_result = a - b
    return print("Sub of a and b s ", sub_result)
```

```python
sub(8,9)
```
